# SalesPulse - 2.35 Root Cause Investigation Workflow
## Systematic Root Cause Analysis Pipeline

This notebook demonstrates the end-to-end Root Cause Investigation Workflow for SalesPulse, following a structured 5-task framework to trace metric anomalies (e.g. 50% revenue drops) to true root causes without guessing or premature action.

### Framework Overview:
1. **Task 1: Isolate Time Window** - Detect anomaly date, zoom into hourly breakdown, isolate worst hour.
2. **Task 2: Segment Analysis** - Categorical breakdowns across Customer Type, Payment Method, and Geography.
3. **Task 3: Correlation Analysis** - Contingency tables, Chi-Square tests, and error log profiling.
4. **Task 4: Documentation and Hypothesis** - Formal Investigation Report, root cause classification, and ROI.
5. **Task 5: Validation of Hypothesis** - External telemetry alignment, segment cross-validation, and final verdict.
